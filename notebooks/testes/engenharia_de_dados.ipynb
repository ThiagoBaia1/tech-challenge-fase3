```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_parquet('../data/dados_modelo/municipio_agregado_outras_fontes.parquet')
df.head()
```

```python
df.describe()
```

```python
# Tivemos a ideia de criar uma feature não-linear

inse = df['MEDIA_INSE']
analf = df['indice_analf']
    
# Criar a nova variável não-linear
df['razao_inse_analf'] = np.divide(inse, analf + 1e-5)
```

```python
# Precisamos entender o desvio socioeconômico local
df['raz_alfab_mun_uf'] = np.divide(df['taxa_alfabetizacao_mun'], df['taxa_alfabetizacao_uf'])
```

```python
df['razao_taxa_uf_inse'] = np.divide(df['taxa_alfabetizacao_uf'],df['MEDIA_INSE']+0.001)
```

```python
df.to_parquet('../data/dados_modelo/municipio_eng_feat.parquet')
```
